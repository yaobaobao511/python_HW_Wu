```
def collect_scores():
  print("我是function ming collect scores()")
  scores_array = []

  while True:
    user_input = input("請輸入學生成績（輸入 'q' 結束）：")
    if user_input == 'q':
      break
    try:
      score = int(user_input)
      scores_array.append(score)
    except ValueError:
      print("請輸入有效的數字或 'q'！")

  return scores_array

def print_scores(data_list):
  print("成绩清单 === , 我是function print_scores")
  for i in range(len(data_list)):
    print(f"第 {i+1} 为學生的成績：{data_list[i]} 分")

def calculate_average(data_list):
  if len(data_list) == 0:
    print("没有成绩可以计算平均")
    return

  total = sum(data_list)
  count = len(data_list)
  average = total / count

  print("-"* 20)
  print(f"全班总分：{total}分")
  print(f"全班平均：{average}分")

math_scores = collect_scores()
print_scores(math_scores)
calculate_average(math_scores)
```
